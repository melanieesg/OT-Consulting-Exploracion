```python
import pandas as pd

colaboradores = pd.read_excel("Catalogo_Colaboradores.xlsx")
df = pd.DataFrame(colaboradores)

def crearcolumnas():
    df["SalarioDiario"] = df["Sueldo Neto Total"] / 192

    df["Compensacion"] = (
        df["Sueldo Asimilados/Sindicato"] +
        df["Sueldo Complementos"]
    )

    df["CostoInterno"] = (
        df["Sueldo Neto IMSS"] + df["Compensacion"]
    )

def limpieza():
    df.drop(columns=["RFC", "CURP", "NSS"], inplace=True)

def estadisticasreferencia():
    print("Salario diario promedio:", df["SalarioDiario"].mean())
    print("Sueldo IMSS promedio:", df["Sueldo Neto IMSS"].mean())
    print("Complementos promedio:", df["Sueldo Complementos"].mean())
    print("Costo interno promedio:", df["CostoInterno"].mean())

def actualiza():
    df.to_excel("Colaboradores_Actualizado.xlsx", index=False)

crearcolumnas()
limpieza()
estadisticasreferencia()
actualiza()
```
